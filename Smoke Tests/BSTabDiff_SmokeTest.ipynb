# Plain default BSTabDiff Smoke Test

```python
import os
# Suppress Python warnings before importing third-party libraries.
os.environ["PYTHONWARNINGS"] = "ignore"
import warnings
warnings.simplefilter("ignore")
import numpy as np
import torch
from bstabdiff import BSTabDiff

# ============================================================
# Synthetic smoke-test data
# ============================================================

rng = np.random.default_rng(0)
n, m = 20, 100
X = rng.standard_normal((n, m)).astype(np.float32)
y = rng.integers(0, 2, size=n)

# ============================================================
# Fit BSTabDiff
# ============================================================

model = BSTabDiff(
    M=10,
    prior_type="diffusion",
    device="cuda" if torch.cuda.is_available() else "cpu",
    random_state=0,
    prior_epochs=5,
    prior_batch=16,
    verbose_every=1,
    use_gobs=False,
)

returned = model.fit(X, y)

# ============================================================
# Basic sklearn-style checks
# ============================================================
assert returned is model
assert model.n_features_in_ == m
assert hasattr(model, "generator_")
assert hasattr(model, "train_info_")

# ============================================================
# Generate synthetic samples
# ============================================================

X_syn, R_syn, y_syn = model.sample(
    n_samples=5,
    return_mask=True,
)

# ============================================================
# Shape checks
# ============================================================

assert X_syn.shape == (5, m)
assert R_syn.shape == (5, m)
assert y_syn is not None
assert y_syn.shape == (5,)

print("\n" + "=" * 70)
print("Plain BSTabDiff smoke test OK")
print("=" * 70)
print("Device      :", model.device_)
print("Synthetic X:", X_syn.shape)
print("Synthetic R:", R_syn.shape)
print("Synthetic y:", y_syn.shape)
```

# BSTabDiff default on Colon w/ Logistic Regression

```python
import os, random, warnings
os.environ["PYTHONWARNINGS"] = "ignore"
warnings.simplefilter("ignore")

import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state
from bstabdiff import BSTabDiff

GLOBAL_SEED = 42
CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"
N_SPLITS = 5
N_REPEATS = 5
N_SYN = 200
M = 32
PRIOR_TYPE = "diffusion"
PRIOR_EPOCHS = 10000
PRIOR_BATCH = 128
PRIOR_LR = 1e-3
EMA_DECAY = 0.999
DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

def load_colon_dataset(csv_path=CSV_PATH, label_col=LABEL_COL):
    df = pd.read_csv(csv_path)
    if label_col not in df.columns:
        raise ValueError(f"Label column '{label_col}' not found.")
    y_raw = df[label_col].values
    X_df = df.drop(columns=[label_col]).select_dtypes(include=[np.number])
    X = np.nan_to_num(X_df.to_numpy(dtype=np.float32), nan=0.0, posinf=0.0, neginf=0.0)
    if y_raw.dtype.kind in {"U", "S", "O"}:
        y, _ = pd.factorize(y_raw)
    else:
        unique = np.unique(y_raw)
        y, _ = pd.factorize(y_raw) if not np.array_equal(unique, np.arange(unique.size)) else (y_raw.astype(int), None)
    return X, np.asarray(y, dtype=int), list(X_df.columns)

def make_lr():
    return Pipeline([
        ("imp", SimpleImputer(strategy="median")),
        ("sc", StandardScaler(with_mean=True, with_std=True)),
        ("clf", LogisticRegression(max_iter=10000, class_weight="balanced", solver="lbfgs", n_jobs=None))
    ])

set_seed(GLOBAL_SEED)
X, y, feature_names = load_colon_dataset()
print(f"Loaded Colon dataset: X.shape={X.shape}, y.shape={y.shape}, classes={np.unique(y)}")
assert X.shape == (62, 2000), f"Expected Colon shape (62, 2000), got {X.shape}"

rng = check_random_state(GLOBAL_SEED)
rkf = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=GLOBAL_SEED
)

tstr_acc, tstr_auc = [], []
trtr_acc, trtr_auc = [], []

for fold_id, (tr, te) in enumerate(rkf.split(X, y), start=1):
    Xtr, ytr = X[tr], y[tr]
    Xte, yte = X[te], y[te]
    fold_seed = int(rng.randint(0, 2**31 - 1))

    print("\n" + "=" * 70)
    print(f"Fold {fold_id:02d}/25 | seed={fold_seed}")
    print("=" * 70)

    model = BSTabDiff(
        M=M,
        prior_type=PRIOR_TYPE,
        device=DEVICE,
        random_state=fold_seed,
        prior_epochs=PRIOR_EPOCHS,
        prior_batch=PRIOR_BATCH,
        prior_lr=PRIOR_LR,
        save_dir=None,
        save_name="bstabdiff_colon",
        save_best=True,
        use_ema=True,
        ema_decay=EMA_DECAY,
        verbose_every=0,
        use_gobs=False,
        use_gobs_fc=False
    )

    model.fit(Xtr, ytr)
    X_syn, R_syn, y_syn = model.sample(n_samples=N_SYN, return_mask=True)

    assert X_syn.shape == (N_SYN, X.shape[1])
    assert R_syn.shape == X_syn.shape
    assert y_syn.shape == (N_SYN,)

    clf_tstr = make_lr()
    clf_tstr.fit(X_syn, y_syn)
    proba_tstr = clf_tstr.predict_proba(Xte)[:, 1]
    pred_tstr = (proba_tstr >= 0.5).astype(int)
    fold_tstr_acc = accuracy_score(yte, pred_tstr)
    fold_tstr_auc = roc_auc_score(yte, proba_tstr)
    tstr_acc.append(fold_tstr_acc)
    tstr_auc.append(fold_tstr_auc)

    clf_trtr = make_lr()
    clf_trtr.fit(Xtr, ytr)
    proba_trtr = clf_trtr.predict_proba(Xte)[:, 1]
    pred_trtr = (proba_trtr >= 0.5).astype(int)
    fold_trtr_acc = accuracy_score(yte, pred_trtr)
    fold_trtr_auc = roc_auc_score(yte, proba_trtr)
    trtr_acc.append(fold_trtr_acc)
    trtr_auc.append(fold_trtr_auc)

    print(f"TSTR | ACC={fold_tstr_acc:.4f} | AUC={fold_tstr_auc:.4f}")
    print(f"TRTR | ACC={fold_trtr_acc:.4f} | AUC={fold_trtr_auc:.4f}")

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

tstr_acc = np.asarray(tstr_acc)
tstr_auc = np.asarray(tstr_auc)
trtr_acc = np.asarray(trtr_acc)
trtr_auc = np.asarray(trtr_auc)

print("\n\n" + "=" * 70)
print("COLON — DEFAULT BSTabDiff + Logistic Regression")
print("5 x 5 repeated stratified CV")
print("=" * 70)
print("\nTSTR:")
print(f"ACC = {tstr_acc.mean():.4f} ± {tstr_acc.std():.4f}")
print(f"AUC = {tstr_auc.mean():.4f} ± {tstr_auc.std():.4f}")
print("\nTRTR:")
print(f"ACC = {trtr_acc.mean():.4f} ± {trtr_acc.std():.4f}")
print(f"AUC = {trtr_auc.mean():.4f} ± {trtr_auc.std():.4f}")
```

```python
import os, random, warnings
os.environ["PYTHONWARNINGS"] = "ignore"
warnings.simplefilter("ignore")

import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state
from bstabdiff import BSTabDiff

GLOBAL_SEED = 42
CV_SEED = 0
CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"
N_SPLITS = 5
N_REPEATS = 5
N_SYN = 200
M = 32
PRIOR_TYPE = "diffusion"
PRIOR_EPOCHS = 10000
PRIOR_BATCH = 128
PRIOR_LR = 1e-3
EMA_DECAY = 0.999
DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

def load_colon_dataset(csv_path=CSV_PATH, label_col=LABEL_COL):
    df = pd.read_csv(csv_path)
    if label_col not in df.columns:
        raise ValueError(f"Label column '{label_col}' not found. Columns: {list(df.columns)[:10]} ...")
    y_raw = df[label_col].values
    X_df = df.drop(columns=[label_col]).select_dtypes(include=[np.number])
    X = np.nan_to_num(X_df.to_numpy(dtype=np.float32), nan=0.0, posinf=0.0, neginf=0.0)
    if y_raw.dtype.kind in {"U", "S", "O"}:
        y, _ = pd.factorize(y_raw)
    else:
        unique = np.unique(y_raw)
        y, _ = pd.factorize(y_raw) if not np.array_equal(unique, np.arange(unique.size)) else (y_raw.astype(int), None)
    return X, np.asarray(y, dtype=int), list(X_df.columns)

def make_lr():
    return Pipeline([
        ("imp", SimpleImputer(strategy="median")),
        ("sc", StandardScaler(with_mean=True, with_std=True)),
        ("clf", LogisticRegression(max_iter=10000, class_weight="balanced", solver="lbfgs", n_jobs=None))
    ])

set_seed(GLOBAL_SEED)
X, y, feature_names = load_colon_dataset()
print(f"Loaded Colon dataset: X.shape={X.shape}, y.shape={y.shape}, classes={np.unique(y)}")
assert X.shape == (62, 2000), f"Expected Colon shape (62, 2000), got {X.shape}"

rng = check_random_state(CV_SEED)
rkf = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=CV_SEED
)

tstr_acc, tstr_auc = [], []
trtr_acc, trtr_auc = [], []
fold_seeds = []

for fold_id, (tr, te) in enumerate(rkf.split(X, y), start=1):
    Xtr, ytr = X[tr], y[tr]
    Xte, yte = X[te], y[te]
    fold_seed = int(rng.randint(0, 2**31 - 1))
    fold_seeds.append(fold_seed)

    print("\n" + "=" * 72)
    print(f"Fold {fold_id:02d}/25 | train={len(tr)} | test={len(te)} | generator_seed={fold_seed}")
    print("=" * 72)

    model = BSTabDiff(
        feature_specs=None,
        M=M,
        blocks=None,
        permute_features=False,
        prior_type=PRIOR_TYPE,
        device=DEVICE,
        random_state=fold_seed,
        prior_epochs=PRIOR_EPOCHS,
        prior_batch=PRIOR_BATCH,
        prior_lr=PRIOR_LR,
        verbose_every=0,
        save_dir=None,
        save_name="bstabdiff_colon",
        save_best=True,
        use_ema=True,
        ema_decay=EMA_DECAY,
        use_gobs=False,
        use_gobs_fc=False
    )

    model.fit(Xtr, ytr)
    X_syn, R_syn, y_syn = model.sample(n_samples=N_SYN, return_mask=True)

    assert X_syn.shape == (N_SYN, X.shape[1])
    assert R_syn.shape == (N_SYN, X.shape[1])
    assert y_syn is not None
    assert y_syn.shape == (N_SYN,)

    clf_tstr = make_lr()
    clf_tstr.fit(X_syn, y_syn)
    proba_tstr = clf_tstr.predict_proba(Xte)[:, 1]
    pred_tstr = (proba_tstr >= 0.5).astype(int)
    fold_tstr_acc = accuracy_score(yte, pred_tstr)
    fold_tstr_auc = roc_auc_score(yte, proba_tstr)
    tstr_acc.append(fold_tstr_acc)
    tstr_auc.append(fold_tstr_auc)

    clf_trtr = make_lr()
    clf_trtr.fit(Xtr, ytr)
    proba_trtr = clf_trtr.predict_proba(Xte)[:, 1]
    pred_trtr = (proba_trtr >= 0.5).astype(int)
    fold_trtr_acc = accuracy_score(yte, pred_trtr)
    fold_trtr_auc = roc_auc_score(yte, proba_trtr)
    trtr_acc.append(fold_trtr_acc)
    trtr_auc.append(fold_trtr_auc)

    print(f"TSTR | ACC={fold_tstr_acc:.4f} | AUC={fold_tstr_auc:.4f}")
    print(f"TRTR | ACC={fold_trtr_acc:.4f} | AUC={fold_trtr_auc:.4f}")

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

tstr_acc = np.asarray(tstr_acc, dtype=float)
tstr_auc = np.asarray(tstr_auc, dtype=float)
trtr_acc = np.asarray(trtr_acc, dtype=float)
trtr_auc = np.asarray(trtr_auc, dtype=float)

print("\n\n" + "=" * 72)
print("COLON — DEFAULT BSTabDiff + Logistic Regression")
print("STRICT 5 x 5 repeated stratified CV")
print(f"GLOBAL_SEED = {GLOBAL_SEED}")
print(f"CV_SEED     = {CV_SEED}")
print(f"BSTabDiff   = M={M}, {PRIOR_TYPE}, epochs={PRIOR_EPOCHS}, batch={PRIOR_BATCH}, lr={PRIOR_LR}, EMA={EMA_DECAY}")
print(f"Device      = {DEVICE}")
print("=" * 72)

print("\nTSTR:")
print(f"ACC = {tstr_acc.mean():.4f} ± {tstr_acc.std():.4f}")
print(f"AUC = {tstr_auc.mean():.4f} ± {tstr_auc.std():.4f}")

print("\nTRTR:")
print(f"ACC = {trtr_acc.mean():.4f} ± {trtr_acc.std():.4f}")
print(f"AUC = {trtr_auc.mean():.4f} ± {trtr_auc.std():.4f}")

print("\n" + "-" * 72)
print("Original Titan notebook reference")
print("-" * 72)
print("TSTR ACC = 0.8326 ± 0.1240")
print("TSTR AUC = 0.8847 ± 0.0922")
print("TRTR ACC = 0.8613 ± 0.0856")
print("TRTR AUC = 0.8745 ± 0.1056")

print("\n" + "-" * 72)
print("Difference: current run - original Titan run")
print("-" * 72)
print(f"TSTR ACC Δ = {tstr_acc.mean() - 0.8326:+.4f}")
print(f"TSTR AUC Δ = {tstr_auc.mean() - 0.8847:+.4f}")
print(f"TRTR ACC Δ = {trtr_acc.mean() - 0.8613:+.4f}")
print(f"TRTR AUC Δ = {trtr_auc.mean() - 0.8745:+.4f}")

print("\nFold generator seeds:")
print(fold_seeds)
```

# w/ GO-BS and GO-BS-FC Smoke test

```python
import os, warnings
os.environ["PYTHONWARNINGS"] = "ignore"
warnings.simplefilter("ignore")

import numpy as np
import torch
from sklearn.base import clone
from bstabdiff import BSTabDiff

SEED = 0
N = 20
M_FEATURES = 100
N_BLOCKS = 10
N_SYN = 5
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

rng = np.random.default_rng(SEED)
X = rng.standard_normal((N, M_FEATURES)).astype(np.float32)
y = rng.integers(0, 2, size=N)

print("=" * 72)
print("BSTabDiff GO-BS / GO-BS-FC smoke tests")
print("=" * 72)
print("Device :", DEVICE)
print("X shape:", X.shape)
print("y shape:", y.shape)

def run_smoke_test(name, use_gobs, use_gobs_fc):
    print("\n" + "=" * 72)
    print(name)
    print("=" * 72)

    model = BSTabDiff(
        M=N_BLOCKS,
        prior_type="diffusion",
        device=DEVICE,
        random_state=SEED,
        prior_epochs=5,
        prior_batch=16,
        prior_lr=1e-3,
        verbose_every=1,
        save_dir=None,
        save_best=True,
        use_ema=True,
        ema_decay=0.999,
        use_gobs=use_gobs,
        use_gobs_fc=use_gobs_fc,
        gobs_num_clusters=3,
        gobs_metric="kl_divergence",
        gobs_bins=32,
        gobs_top_k=None,
        gobs_refine_order=True,
        gobs_direction_select=True,
        gobs_refine_passes=1,
        gobs_boundary_refine_passes=5,
        gobs_boundary_window=8,
        gobs_lambda_cross=1.0,
        gobs_gamma_balance=1e-3,
        gobs_use_cpu_kmeans=False
    )

    returned = model.fit(X, y)

    assert returned is model
    assert hasattr(model, "generator_")
    assert hasattr(model, "train_info_")
    assert model.n_features_in_ == M_FEATURES

    X_syn, R_syn, y_syn = model.sample(
        n_samples=N_SYN,
        return_mask=True
    )

    assert X_syn.shape == (N_SYN, M_FEATURES)
    assert R_syn.shape == (N_SYN, M_FEATURES)
    assert y_syn is not None
    assert y_syn.shape == (N_SYN,)
    assert np.isfinite(X_syn[R_syn.astype(bool)]).all()

    params = model.get_params()
    assert params["M"] == N_BLOCKS
    assert params["use_gobs"] == use_gobs
    assert params["use_gobs_fc"] == use_gobs_fc

    cloned = clone(model)
    assert cloned.M == model.M
    assert cloned.use_gobs == model.use_gobs
    assert cloned.use_gobs_fc == model.use_gobs_fc

    print("Smoke test OK")
    print("Synthetic X :", X_syn.shape)
    print("Synthetic R :", R_syn.shape)
    print("Synthetic y :", y_syn.shape)
    print("Observed frac:", f"{R_syn.mean():.6f}")
    print("use_gobs     :", model.use_gobs)
    print("use_gobs_fc  :", model.use_gobs_fc)
    print("Device       :", model.device_)

    return model

model_gobs = run_smoke_test(
    name="BSTabDiff + GO-BS",
    use_gobs=True,
    use_gobs_fc=False
)

if torch.cuda.is_available():
    torch.cuda.empty_cache()

model_gobs_fc = run_smoke_test(
    name="BSTabDiff + GO-BS-FC",
    use_gobs=True,
    use_gobs_fc=True
)

print("\n" + "=" * 72)
print("ALL BSTabDiff ordering smoke tests PASSED")
print("=" * 72)
print("✓ GO-BS")
print("✓ GO-BS-FC")
print("✓ sklearn fit()")
print("✓ sample()")
print("✓ return_mask=True")
print("✓ get_params()")
print("✓ clone()")
```

# w/ GO-BS and GO-BS-FC on Colon

```python
import os
os.environ['PYTHONWARNINGS'] = 'ignore'
import warnings
warnings.simplefilter('ignore')
import random
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state
from bstabdiff import BSTabDiff
GLOBAL_SEED = 42
CV_SEED = 0
CSV_PATH = 'coloncancer_encoded.csv'
LABEL_COL = 'label'
N_SPLITS = 5
N_REPEATS = 5
DEVICE = 'cuda:0' if torch.cuda.is_available() else 'cpu'
M = 8
PRIOR_TYPE = 'flow'
PRIOR_EPOCHS = 300
PRIOR_BATCH = 64
PRIOR_LR = 0.0032931
EMA_DECAY = 0.9926
N_SYN = 64
GOBS_NUM_CLUSTERS = 3
GOBS_METRIC = 'correlation'
GOBS_BINS = 24
GOBS_TOP_K = 16
GOBS_REFINE_ORDER = False
GOBS_DIRECTION_SELECT = False
GOBS_REFINE_PASSES = 3
GOBS_BOUNDARY_REFINE_PASSES = 5
GOBS_BOUNDARY_WINDOW = 16
GOBS_LAMBDA_CROSS = 2.3989
GOBS_GAMMA_BALANCE = 5.19e-06
PAPER_RESULTS = {'GO-BS': {'tstr_acc': 0.8287, 'tstr_auc': 0.8878, 'tstr_acc_std': 0.0994, 'tstr_auc_std': 0.0961}, 'GO-BS-FC': {'tstr_acc': 0.8321, 'tstr_auc': 0.8935, 'tstr_acc_std': 0.1009, 'tstr_auc_std': 0.0917}, 'TRTR': {'acc': 0.8613, 'auc': 0.8745, 'acc_std': 0.0856, 'auc_std': 0.1056}}

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
set_seed(GLOBAL_SEED)

def load_colon_dataset(csv_path=CSV_PATH, label_col=LABEL_COL):
    df = pd.read_csv(csv_path)
    if label_col not in df.columns:
        raise ValueError(f"Label column '{label_col}' not found. Columns: {list(df.columns)[:10]} ...")
    y_raw = df[label_col].values
    X_df = df.drop(columns=[label_col])
    X_df = X_df.select_dtypes(include=[np.number])
    X = X_df.to_numpy(dtype=np.float32)
    X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)
    if y_raw.dtype.kind in {'U', 'S', 'O'}:
        y, _ = pd.factorize(y_raw)
    else:
        unique = np.unique(y_raw)
        if not np.array_equal(unique, np.arange(unique.size)):
            y, _ = pd.factorize(y_raw)
        else:
            y = y_raw.astype(int, copy=False)
    return (X, np.asarray(y, dtype=int), list(X_df.columns))
X, y, feature_names = load_colon_dataset()
print('=' * 76)
print(f'Loaded Colon: X.shape={X.shape}, y.shape={y.shape}, classes={np.unique(y)}')
print(f'Device: {DEVICE}')
print('=' * 76)
assert X.shape == (62, 2000), f'Expected Colon shape (62, 2000), got {X.shape}'

def make_lr():
    return Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler(with_mean=True, with_std=True)), ('clf', LogisticRegression(max_iter=10000, class_weight='balanced', solver='lbfgs', n_jobs=None))])

def evaluate_ordering(method_name, use_gobs_fc):
    """

    Strict 5x5 repeated stratified CV.

    GO-BS:

        use_gobs=True

        use_gobs_fc=False

    GO-BS-FC:

        use_gobs=True

        use_gobs_fc=True

    Both methods use identical:

        - folds

        - fold seeds

        - BSTabDiff hyperparameters

        - GO-BS hyperparameters

        - Logistic Regression evaluator

    """
    set_seed(GLOBAL_SEED)
    rng = check_random_state(CV_SEED)
    rkf = RepeatedStratifiedKFold(n_splits=N_SPLITS, n_repeats=N_REPEATS, random_state=CV_SEED)
    tstr_acc = []
    tstr_auc = []
    trtr_acc = []
    trtr_auc = []
    fold_seeds = []
    print('\n\n' + '#' * 76)
    print(f'# {method_name}')
    print('#' * 76)
    print(f'M={M} | prior={PRIOR_TYPE} | epochs={PRIOR_EPOCHS} | batch={PRIOR_BATCH} | lr={PRIOR_LR} | EMA={EMA_DECAY} | n_syn={N_SYN}')
    print(f'GO-BS clusters={GOBS_NUM_CLUSTERS} | metric={GOBS_METRIC} | bins={GOBS_BINS} | top_k={GOBS_TOP_K}')
    print(f'use_gobs=True | use_gobs_fc={use_gobs_fc}')
    print('#' * 76)
    for fold_id, (tr, te) in enumerate(rkf.split(X, y), start=1):
        Xtr = X[tr]
        ytr = y[tr]
        Xte = X[te]
        yte = y[te]
        fold_seed = int(rng.randint(0, 2 ** 31 - 1))
        fold_seeds.append(fold_seed)
        print('\n' + '=' * 76)
        print(f'{method_name} | Fold {fold_id:02d}/25 | train={len(tr)} | test={len(te)} | seed={fold_seed}')
        print('=' * 76)
        model = BSTabDiff(feature_specs=None, M=M, blocks=None, permute_features=False, prior_type=PRIOR_TYPE, device=DEVICE, random_state=fold_seed, prior_epochs=PRIOR_EPOCHS, prior_batch=PRIOR_BATCH, prior_lr=PRIOR_LR, verbose_every=0, save_dir=None, save_name='bstabdiff_colon_gobs_fc' if use_gobs_fc else 'bstabdiff_colon_gobs', save_best=True, use_ema=True, ema_decay=EMA_DECAY, use_gobs=True, use_gobs_fc=use_gobs_fc, gobs_num_clusters=GOBS_NUM_CLUSTERS, gobs_metric=GOBS_METRIC, gobs_bins=GOBS_BINS, gobs_top_k=GOBS_TOP_K, gobs_refine_order=GOBS_REFINE_ORDER, gobs_direction_select=GOBS_DIRECTION_SELECT, gobs_refine_passes=GOBS_REFINE_PASSES, gobs_boundary_refine_passes=GOBS_BOUNDARY_REFINE_PASSES, gobs_boundary_window=GOBS_BOUNDARY_WINDOW, gobs_lambda_cross=GOBS_LAMBDA_CROSS, gobs_gamma_balance=GOBS_GAMMA_BALANCE, gobs_use_cpu_kmeans=False)
        model.fit(Xtr, ytr)
        X_syn, R_syn, y_syn = model.sample(n_samples=N_SYN, return_mask=True)
        assert X_syn.shape == (N_SYN, X.shape[1])
        assert R_syn.shape == (N_SYN, X.shape[1])
        assert y_syn is not None
        assert y_syn.shape == (N_SYN,)
        clf_tstr = make_lr()
        clf_tstr.fit(X_syn, y_syn)
        proba_tstr = clf_tstr.predict_proba(Xte)[:, 1]
        pred_tstr = (proba_tstr >= 0.5).astype(int)
        fold_tstr_acc = accuracy_score(yte, pred_tstr)
        fold_tstr_auc = roc_auc_score(yte, proba_tstr)
        tstr_acc.append(fold_tstr_acc)
        tstr_auc.append(fold_tstr_auc)
        clf_trtr = make_lr()
        clf_trtr.fit(Xtr, ytr)
        proba_trtr = clf_trtr.predict_proba(Xte)[:, 1]
        pred_trtr = (proba_trtr >= 0.5).astype(int)
        fold_trtr_acc = accuracy_score(yte, pred_trtr)
        fold_trtr_auc = roc_auc_score(yte, proba_trtr)
        trtr_acc.append(fold_trtr_acc)
        trtr_auc.append(fold_trtr_auc)
        print(f'TSTR | ACC={fold_tstr_acc:.4f} | AUC={fold_tstr_auc:.4f}')
        print(f'TRTR | ACC={fold_trtr_acc:.4f} | AUC={fold_trtr_auc:.4f}')
        del model
        del clf_tstr
        del clf_trtr
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
    tstr_acc = np.asarray(tstr_acc, dtype=float)
    tstr_auc = np.asarray(tstr_auc, dtype=float)
    trtr_acc = np.asarray(trtr_acc, dtype=float)
    trtr_auc = np.asarray(trtr_auc, dtype=float)
    result = {'method': method_name, 'tstr_acc_mean': tstr_acc.mean(), 'tstr_acc_std': tstr_acc.std(), 'tstr_auc_mean': tstr_auc.mean(), 'tstr_auc_std': tstr_auc.std(), 'trtr_acc_mean': trtr_acc.mean(), 'trtr_acc_std': trtr_acc.std(), 'trtr_auc_mean': trtr_auc.mean(), 'trtr_auc_std': trtr_auc.std(), 'fold_seeds': fold_seeds}
    print('\n\n' + '=' * 76)
    print(f'COLON — {method_name}')
    print('STRICT 5 x 5 repeated stratified CV')
    print(f'GLOBAL_SEED = {GLOBAL_SEED}')
    print(f'CV_SEED     = {CV_SEED}')
    print(f'Device      = {DEVICE}')
    print('=' * 76)
    print('\nTSTR:')
    print(f"ACC = {result['tstr_acc_mean']:.4f} ± {result['tstr_acc_std']:.4f}")
    print(f"AUC = {result['tstr_auc_mean']:.4f} ± {result['tstr_auc_std']:.4f}")
    print('\nTRTR:')
    print(f"ACC = {result['trtr_acc_mean']:.4f} ± {result['trtr_acc_std']:.4f}")
    print(f"AUC = {result['trtr_auc_mean']:.4f} ± {result['trtr_auc_std']:.4f}")
    ref = PAPER_RESULTS[method_name]
    print('\n' + '-' * 76)
    print('Paper reference')
    print('-' * 76)
    print(f"TSTR ACC = {ref['tstr_acc']:.4f} ± {ref['tstr_acc_std']:.4f}")
    print(f"TSTR AUC = {ref['tstr_auc']:.4f} ± {ref['tstr_auc_std']:.4f}")
    print(f"TRTR ACC = {PAPER_RESULTS['TRTR']['acc']:.4f} ± {PAPER_RESULTS['TRTR']['acc_std']:.4f}")
    print(f"TRTR AUC = {PAPER_RESULTS['TRTR']['auc']:.4f} ± {PAPER_RESULTS['TRTR']['auc_std']:.4f}")
    print('\n' + '-' * 76)
    print('Difference: current run - paper')
    print('-' * 76)
    print(f"TSTR ACC Δ = {result['tstr_acc_mean'] - ref['tstr_acc']:+.4f}")
    print(f"TSTR AUC Δ = {result['tstr_auc_mean'] - ref['tstr_auc']:+.4f}")
    print(f"TRTR ACC Δ = {result['trtr_acc_mean'] - PAPER_RESULTS['TRTR']['acc']:+.4f}")
    print(f"TRTR AUC Δ = {result['trtr_auc_mean'] - PAPER_RESULTS['TRTR']['auc']:+.4f}")
    print('\nFold generator seeds:')
    print(fold_seeds)
    return result
result_gobs = evaluate_ordering(method_name='GO-BS', use_gobs_fc=False)
if torch.cuda.is_available():
    torch.cuda.empty_cache()
result_gobs_fc = evaluate_ordering(method_name='GO-BS-FC', use_gobs_fc=True)
summary = pd.DataFrame([{'Method': 'GO-BS', 'TSTR ACC': result_gobs['tstr_acc_mean'], 'TSTR ACC std': result_gobs['tstr_acc_std'], 'TSTR AUC': result_gobs['tstr_auc_mean'], 'TSTR AUC std': result_gobs['tstr_auc_std'], 'TRTR ACC': result_gobs['trtr_acc_mean'], 'TRTR AUC': result_gobs['trtr_auc_mean']}, {'Method': 'GO-BS-FC', 'TSTR ACC': result_gobs_fc['tstr_acc_mean'], 'TSTR ACC std': result_gobs_fc['tstr_acc_std'], 'TSTR AUC': result_gobs_fc['tstr_auc_mean'], 'TSTR AUC std': result_gobs_fc['tstr_auc_std'], 'TRTR ACC': result_gobs_fc['trtr_acc_mean'], 'TRTR AUC': result_gobs_fc['trtr_auc_mean']}])
print('\n\n' + '=' * 90)
print('FINAL COLON GO-BS / GO-BS-FC COMPARISON')
print('=' * 90)
print(summary.to_string(index=False, float_format=lambda x: f'{x:.4f}'))
assert result_gobs['fold_seeds'] == result_gobs_fc['fold_seeds'], 'GO-BS and GO-BS-FC did not use identical generator seeds.'
print('\n✓ GO-BS and GO-BS-FC used identical CV splits and fold generator seeds.')
summary.to_csv('colon_gobs_gobsfc_reproduction.csv', index=False)
print('✓ Saved: colon_gobs_gobsfc_reproduction.csv')
```
