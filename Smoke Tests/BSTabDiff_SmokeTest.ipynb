# Plain default BSTabDiff Smoke Test

```python
import os
# Suppress Python warnings before importing third-party libraries.
os.environ["PYTHONWARNINGS"] = "ignore"
import warnings
warnings.simplefilter("ignore")
import numpy as np
import torch
from bstabdiff import BSTabDiff

# ============================================================
# Synthetic smoke-test data
# ============================================================

rng = np.random.default_rng(0)
n, m = 20, 100
X = rng.standard_normal((n, m)).astype(np.float32)
y = rng.integers(0, 2, size=n)

# ============================================================
# Fit BSTabDiff
# ============================================================

model = BSTabDiff(
    M=10,
    prior_type="diffusion",
    device="cuda" if torch.cuda.is_available() else "cpu",
    random_state=0,
    prior_epochs=5,
    prior_batch=16,
    verbose_every=1,
    use_gobs=False,
)

returned = model.fit(X, y)

# ============================================================
# Basic sklearn-style checks
# ============================================================
assert returned is model
assert model.n_features_in_ == m
assert hasattr(model, "generator_")
assert hasattr(model, "train_info_")

# ============================================================
# Generate synthetic samples
# ============================================================

X_syn, R_syn, y_syn = model.sample(
    n_samples=5,
    return_mask=True,
)

# ============================================================
# Shape checks
# ============================================================

assert X_syn.shape == (5, m)
assert R_syn.shape == (5, m)
assert y_syn is not None
assert y_syn.shape == (5,)

print("\n" + "=" * 70)
print("Plain BSTabDiff smoke test OK")
print("=" * 70)
print("Device      :", model.device_)
print("Synthetic X:", X_syn.shape)
print("Synthetic R:", R_syn.shape)
print("Synthetic y:", y_syn.shape)
```

# BSTabDiff default on Colon w/ Logistic Regression

```python
import os

# ============================================================
# Suppress warnings
# ============================================================

os.environ["PYTHONWARNINGS"] = "ignore"
import warnings
warnings.simplefilter("ignore")


# ============================================================
# Imports
# ============================================================


import random
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state
from bstabdiff import BSTabDiff

# ============================================================
# Configuration
# ============================================================

GLOBAL_SEED = 42

CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"

N_SPLITS = 5
N_REPEATS = 5

# Same synthetic sample size as notebook
N_SYN = 200


# Same BSTabDiff configuration as notebook
M = 32
PRIOR_TYPE = "diffusion"
PRIOR_EPOCHS = 10000
PRIOR_BATCH = 128
PRIOR_LR = 1e-3
EMA_DECAY = 0.999

DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"

# ============================================================
# Reproducibility
# ============================================================


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)


    torch.manual_seed(seed)


    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

set_seed(GLOBAL_SEED)


# ============================================================
# Load Colon dataset
# Same logic as notebook
# ============================================================


def load_colon_dataset(
    csv_path=CSV_PATH,
    label_col=LABEL_COL,
):
    df = pd.read_csv(csv_path)


    if label_col not in df.columns:
        raise ValueError(
            f"Label column '{label_col}' not found."
        )


    y_raw = df[label_col].values


    X_df = df.drop(columns=[label_col])


    # Numeric features only
    X_df = X_df.select_dtypes(include=[np.number])


    X = X_df.to_numpy(dtype=np.float32)


    # Same preprocessing used in notebook
    X = np.nan_to_num(
        X,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )


    # Convert labels to 0,...,C-1
    if y_raw.dtype.kind in {"U", "S", "O"}:
        y, _ = pd.factorize(y_raw)


    else:
        unique = np.unique(y_raw)


        if not np.array_equal(
            unique,
            np.arange(unique.size),
        ):
            y, _ = pd.factorize(y_raw)


        else:
            y = y_raw.astype(
                int,
                copy=False,
            )


    return (
        X,
        np.asarray(y, dtype=int),
        list(X_df.columns),
    )


X, y, feature_names = load_colon_dataset()


print(
    f"Loaded Colon dataset: "
    f"X.shape={X.shape}, "
    f"y.shape={y.shape}, "
    f"classes={np.unique(y)}"
)


assert X.shape == (62, 2000), (
    f"Expected Colon shape (62, 2000), got {X.shape}"
)


# ============================================================
# Logistic Regression
# EXACT configuration from notebook
# ============================================================


def make_lr():
    return Pipeline(
        [
            (
                "imp",
                SimpleImputer(
                    strategy="median",
                ),
            ),


            (
                "sc",
                StandardScaler(
                    with_mean=True,
                    with_std=True,
                ),
            ),


            (
                "clf",
                LogisticRegression(
                    max_iter=10000,
                    class_weight="balanced",
                    solver="lbfgs",
                    n_jobs=None,
                ),
            ),
        ]
    )




# ============================================================
# 5x5 repeated stratified CV
# ============================================================


rng = check_random_state(GLOBAL_SEED)


rkf = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=GLOBAL_SEED,
)




tstr_acc = []
tstr_auc = []


trtr_acc = []
trtr_auc = []




# ============================================================
# Run
# ============================================================


for fold_id, (tr, te) in enumerate(
    rkf.split(X, y),
    start=1,
):


    Xtr = X[tr]
    ytr = y[tr]


    Xte = X[te]
    yte = y[te]


    # Same approach as notebook:
    # generate a different deterministic seed per fold.
    fold_seed = int(
        rng.randint(
            0,
            2**31 - 1,
        )
    )


    print(
        "\n"
        + "=" * 70
    )

    print(
        f"Fold {fold_id:02d}/25 | "
        f"seed={fold_seed}"
    )

    print(
        "=" * 70
    )

    # ========================================================
    # 1. Fit DEFAULT BSTabDiff
    #
    # use_gobs=False:
    # this is the standard/equal-block version.
    # ========================================================

    model = BSTabDiff(
        M=M,
        prior_type=PRIOR_TYPE,
        device=DEVICE,
        random_state=fold_seed,

        prior_epochs=PRIOR_EPOCHS,
        prior_batch=PRIOR_BATCH,
        prior_lr=PRIOR_LR,

        save_dir=None,
        save_name="bstabdiff_colon",
        save_best=True,

        use_ema=True,
        ema_decay=EMA_DECAY,

        verbose_every=0,

        # Important:
        # Default BSTabDiff, NO GO-BS
        use_gobs=False,
        use_gobs_fc=False,
    )

    # ========================================================
    # Train generator ONLY on fold training data
    # ========================================================


    model.fit(
        Xtr,
        ytr,
    )

    # ========================================================
    # Generate 200 synthetic samples
    # ========================================================


    X_syn, R_syn, y_syn = model.sample(
        n_samples=N_SYN,
        return_mask=True,
    )

    assert X_syn.shape == (
        N_SYN,
        X.shape[1],
    )

    assert R_syn.shape == X_syn.shape


    assert y_syn.shape == (
        N_SYN,
    )

    # ========================================================
    # 2. TSTR
    #
    # Train Logistic Regression on SYNTHETIC
    # Test on REAL held-out fold
    # ========================================================


    clf_tstr = make_lr()


    clf_tstr.fit(
        X_syn,
        y_syn,
    )


    proba_tstr = clf_tstr.predict_proba(
        Xte
    )[:, 1]


    pred_tstr = (
        proba_tstr >= 0.5
    ).astype(int)


    fold_tstr_acc = accuracy_score(
        yte,
        pred_tstr,
    )


    fold_tstr_auc = roc_auc_score(
        yte,
        proba_tstr,
    )


    tstr_acc.append(
        fold_tstr_acc
    )


    tstr_auc.append(
        fold_tstr_auc
    )

    # ========================================================
    # 3. TRTR
    #
    # Train Logistic Regression on REAL training fold
    # Test on REAL held-out fold
    #
    # BSTabDiff is NOT involved here.
    # ========================================================


    clf_trtr = make_lr()


    clf_trtr.fit(
        Xtr,
        ytr,
    )


    proba_trtr = clf_trtr.predict_proba(
        Xte
    )[:, 1]


    pred_trtr = (
        proba_trtr >= 0.5
    ).astype(int)


    fold_trtr_acc = accuracy_score(
        yte,
        pred_trtr,
    )


    fold_trtr_auc = roc_auc_score(
        yte,
        proba_trtr,
    )


    trtr_acc.append(
        fold_trtr_acc
    )


    trtr_auc.append(
        fold_trtr_auc
    )

    # ========================================================
    # Report fold
    # ========================================================


    print(
        f"TSTR | "
        f"ACC={fold_tstr_acc:.4f} | "
        f"AUC={fold_tstr_auc:.4f}"
    )


    print(
        f"TRTR | "
        f"ACC={fold_trtr_acc:.4f} | "
        f"AUC={fold_trtr_auc:.4f}"
    )

    # Optional cleanup between generator fits
    del model


    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ============================================================
# Summary
# ============================================================


tstr_acc = np.asarray(tstr_acc)
tstr_auc = np.asarray(tstr_auc)


trtr_acc = np.asarray(trtr_acc)
trtr_auc = np.asarray(trtr_auc)

print(
    "\n\n"
    + "=" * 70
)


print(
    "COLON — DEFAULT BSTabDiff + Logistic Regression"
)


print(
    "5 x 5 repeated stratified CV"
)


print(
    "=" * 70
)


print(
    "\nTSTR:"
)


print(
    f"ACC = "
    f"{tstr_acc.mean():.4f} "
    f"± {tstr_acc.std():.4f}"
)

print(
    f"AUC = "
    f"{tstr_auc.mean():.4f} "
    f"± {tstr_auc.std():.4f}"
)

print(
    "\nTRTR:"
)

print(
    f"ACC = "
    f"{trtr_acc.mean():.4f} "
    f"± {trtr_acc.std():.4f}"
)

print(
    f"AUC = "
    f"{trtr_auc.mean():.4f} "
    f"± {trtr_auc.std():.4f}"
)
```

# The new sklearn-style BSTabDiff.fit()/sample() interface reproduces the original Colon experiment results reported in the paper within expected cross-hardware numerical variation

```python
import os


# ============================================================
# Suppress warnings
# ============================================================


os.environ["PYTHONWARNINGS"] = "ignore"


import warnings
warnings.simplefilter("ignore")




# ============================================================
# Imports
# ============================================================


import random
import numpy as np
import pandas as pd
import torch


from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state


from bstabdiff import BSTabDiff




# ============================================================
# Exact notebook configuration
# ============================================================


# General notebook seed.
# The original notebook calls set_seed(GLOBAL_SEED) before the
# experiment, but the strict 5x5 evaluation itself uses seed=0.
GLOBAL_SEED = 42


# IMPORTANT:
# Exact seed supplied to eval_5x5_tstr_foldwise_gen(...) in
# the original notebook.
CV_SEED = 0




CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"




# Strict evaluation
N_SPLITS = 5
N_REPEATS = 5


# Same number of synthetic samples per fold
N_SYN = 200




# ============================================================
# Exact BSTabDiff configuration from notebook
# ============================================================


M = 32


PRIOR_TYPE = "diffusion"


PRIOR_EPOCHS = 10000
PRIOR_BATCH = 128
PRIOR_LR = 1e-3


EMA_DECAY = 0.999


DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"




# ============================================================
# Reproducibility
# ============================================================


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)


    torch.manual_seed(seed)


    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)




# Same as notebook main block
set_seed(GLOBAL_SEED)




# ============================================================
# Load Colon dataset
# Exact notebook logic
# ============================================================


def load_colon_dataset(
    csv_path=CSV_PATH,
    label_col=LABEL_COL,
):
    df = pd.read_csv(csv_path)


    if label_col not in df.columns:
        raise ValueError(
            f"Label column '{label_col}' not found. "
            f"Columns: {list(df.columns)[:10]} ..."
        )


    y_raw = df[label_col].values


    X_df = df.drop(columns=[label_col])


    # Numeric features only
    X_df = X_df.select_dtypes(include=[np.number])


    X = X_df.to_numpy(dtype=np.float32)


    # Same notebook preprocessing
    X = np.nan_to_num(
        X,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )


    # Make labels 0,...,C-1
    if y_raw.dtype.kind in {"U", "S", "O"}:
        y, _ = pd.factorize(y_raw)


    else:
        unique = np.unique(y_raw)


        if not np.array_equal(
            unique,
            np.arange(unique.size),
        ):
            y, _ = pd.factorize(y_raw)


        else:
            y = y_raw.astype(
                int,
                copy=False,
            )


    return (
        X,
        np.asarray(y, dtype=int),
        list(X_df.columns),
    )




# ============================================================
# Load data
# ============================================================


X, y, feature_names = load_colon_dataset(
    csv_path=CSV_PATH,
    label_col=LABEL_COL,
)


print(
    f"Loaded Colon dataset: "
    f"X.shape={X.shape}, "
    f"y.shape={y.shape}, "
    f"classes={np.unique(y)}"
)


assert X.shape == (62, 2000), (
    f"Expected Colon shape (62, 2000), got {X.shape}"
)




# ============================================================
# Logistic Regression
# EXACT notebook configuration
# ============================================================


def make_lr():
    return Pipeline(
        [
            (
                "imp",
                SimpleImputer(
                    strategy="median",
                ),
            ),


            (
                "sc",
                StandardScaler(
                    with_mean=True,
                    with_std=True,
                ),
            ),


            (
                "clf",
                LogisticRegression(
                    max_iter=10000,
                    class_weight="balanced",
                    solver="lbfgs",
                    n_jobs=None,
                ),
            ),
        ]
    )




# ============================================================
# Strict 5x5 repeated stratified CV
#
# IMPORTANT:
# Original notebook calls:
#
# eval_5x5_tstr_foldwise_gen(..., seed=0)
#
# Therefore BOTH the splitter and fold-seed RNG must use 0.
# ============================================================


rng = check_random_state(CV_SEED)


rkf = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=CV_SEED,
)




# ============================================================
# Metric storage
# ============================================================


tstr_acc = []
tstr_auc = []


trtr_acc = []
trtr_auc = []




# Optional: store seeds so we can verify/reproduce every fold
fold_seeds = []




# ============================================================
# Run strict fold-wise evaluation
# ============================================================


for fold_id, (tr, te) in enumerate(
    rkf.split(X, y),
    start=1,
):


    # --------------------------------------------------------
    # Real train/test fold
    # --------------------------------------------------------


    Xtr = X[tr]
    ytr = y[tr]


    Xte = X[te]
    yte = y[te]




    # --------------------------------------------------------
    # EXACT notebook fold seed generation
    # --------------------------------------------------------


    fold_seed = int(
        rng.randint(
            0,
            2**31 - 1,
        )
    )


    fold_seeds.append(fold_seed)




    print("\n" + "=" * 72)


    print(
        f"Fold {fold_id:02d}/25 | "
        f"train={len(tr)} | "
        f"test={len(te)} | "
        f"generator_seed={fold_seed}"
    )


    print("=" * 72)




    # ========================================================
    # 1. Fit DEFAULT BSTabDiff
    #
    # Exact equivalent of notebook:
    #
    # M=32
    # blocks=None
    # permute_features=False
    # prior_type="diffusion"
    # prior_epochs=10000
    # prior_batch=128
    # prior_lr=1e-3
    # save_dir=None
    # save_name="bstabdiff_colon"
    # save_best=True
    # use_ema=True
    # ema_decay=0.999
    #
    # GO-BS did not exist in this default run, therefore:
    # use_gobs=False
    # ========================================================


    model = BSTabDiff(
        feature_specs=None,


        M=M,


        # Same as blocks=None
        blocks=None,


        # Same notebook setting
        permute_features=False,


        prior_type=PRIOR_TYPE,


        device=DEVICE,


        # Exact fold-specific seed
        random_state=fold_seed,


        prior_epochs=PRIOR_EPOCHS,
        prior_batch=PRIOR_BATCH,
        prior_lr=PRIOR_LR,


        verbose_every=0,


        save_dir=None,
        save_name="bstabdiff_colon",
        save_best=True,


        use_ema=True,
        ema_decay=EMA_DECAY,


        # ----------------------------------------------------
        # Default BSTabDiff — NO GO-BS
        # ----------------------------------------------------
        use_gobs=False,
        use_gobs_fc=False,
    )




    # ========================================================
    # Train generator on TRAINING FOLD ONLY
    # ========================================================


    model.fit(
        Xtr,
        ytr,
    )




    # ========================================================
    # Generate exactly 200 synthetic samples
    # ========================================================


    X_syn, R_syn, y_syn = model.sample(
        n_samples=N_SYN,
        return_mask=True,
    )




    # --------------------------------------------------------
    # Sanity checks
    # --------------------------------------------------------


    assert X_syn.shape == (
        N_SYN,
        X.shape[1],
    )


    assert R_syn.shape == (
        N_SYN,
        X.shape[1],
    )


    assert y_syn is not None


    assert y_syn.shape == (
        N_SYN,
    )




    # ========================================================
    # 2. TSTR
    #
    # Train on Synthetic
    # Test on Real
    # ========================================================


    clf_tstr = make_lr()


    clf_tstr.fit(
        X_syn,
        y_syn,
    )




    proba_tstr = clf_tstr.predict_proba(
        Xte
    )[:, 1]




    pred_tstr = (
        proba_tstr >= 0.5
    ).astype(int)




    fold_tstr_acc = accuracy_score(
        yte,
        pred_tstr,
    )


    fold_tstr_auc = roc_auc_score(
        yte,
        proba_tstr,
    )




    tstr_acc.append(
        fold_tstr_acc
    )


    tstr_auc.append(
        fold_tstr_auc
    )




    # ========================================================
    # 3. TRTR
    #
    # Train on Real
    # Test on Real
    #
    # BSTabDiff has NO involvement in this baseline.
    # ========================================================


    clf_trtr = make_lr()


    clf_trtr.fit(
        Xtr,
        ytr,
    )




    proba_trtr = clf_trtr.predict_proba(
        Xte
    )[:, 1]




    pred_trtr = (
        proba_trtr >= 0.5
    ).astype(int)




    fold_trtr_acc = accuracy_score(
        yte,
        pred_trtr,
    )


    fold_trtr_auc = roc_auc_score(
        yte,
        proba_trtr,
    )




    trtr_acc.append(
        fold_trtr_acc
    )


    trtr_auc.append(
        fold_trtr_auc
    )




    # ========================================================
    # Report fold
    # ========================================================


    print(
        f"TSTR | "
        f"ACC={fold_tstr_acc:.4f} | "
        f"AUC={fold_tstr_auc:.4f}"
    )


    print(
        f"TRTR | "
        f"ACC={fold_trtr_acc:.4f} | "
        f"AUC={fold_trtr_auc:.4f}"
    )




    # ========================================================
    # Cleanup
    # ========================================================


    del model


    if torch.cuda.is_available():
        torch.cuda.empty_cache()




# ============================================================
# Convert results to arrays
# ============================================================


tstr_acc = np.asarray(
    tstr_acc,
    dtype=float,
)


tstr_auc = np.asarray(
    tstr_auc,
    dtype=float,
)


trtr_acc = np.asarray(
    trtr_acc,
    dtype=float,
)


trtr_auc = np.asarray(
    trtr_auc,
    dtype=float,
)




# ============================================================
# Final summary
# ============================================================


print("\n\n" + "=" * 72)


print(
    "COLON — DEFAULT BSTabDiff + Logistic Regression"
)


print(
    "STRICT 5 x 5 repeated stratified CV"
)


print(
    f"GLOBAL_SEED = {GLOBAL_SEED}"
)


print(
    f"CV_SEED     = {CV_SEED}"
)


print(
    f"BSTabDiff   = M={M}, "
    f"{PRIOR_TYPE}, "
    f"epochs={PRIOR_EPOCHS}, "
    f"batch={PRIOR_BATCH}, "
    f"lr={PRIOR_LR}, "
    f"EMA={EMA_DECAY}"
)


print(
    f"Device      = {DEVICE}"
)


print("=" * 72)




# ============================================================
# TSTR
# ============================================================


print("\nTSTR:")


print(
    f"ACC = "
    f"{tstr_acc.mean():.4f} "
    f"± {tstr_acc.std():.4f}"
)


print(
    f"AUC = "
    f"{tstr_auc.mean():.4f} "
    f"± {tstr_auc.std():.4f}"
)




# ============================================================
# TRTR
# ============================================================


print("\nTRTR:")


print(
    f"ACC = "
    f"{trtr_acc.mean():.4f} "
    f"± {trtr_acc.std():.4f}"
)


print(
    f"AUC = "
    f"{trtr_auc.mean():.4f} "
    f"± {trtr_auc.std():.4f}"
)




# ============================================================
# Original notebook reference
# ============================================================


print("\n" + "-" * 72)
print("Original Titan notebook reference")
print("-" * 72)


print(
    "TSTR ACC = 0.8326 ± 0.1240"
)


print(
    "TSTR AUC = 0.8847 ± 0.0922"
)


print(
    "TRTR ACC = 0.8613 ± 0.0856"
)


print(
    "TRTR AUC = 0.8745 ± 0.1056"
)




# ============================================================
# Difference from original notebook
# ============================================================


print("\n" + "-" * 72)
print("Difference: current run - original Titan run")
print("-" * 72)


print(
    f"TSTR ACC Δ = "
    f"{tstr_acc.mean() - 0.8326:+.4f}"
)


print(
    f"TSTR AUC Δ = "
    f"{tstr_auc.mean() - 0.8847:+.4f}"
)


print(
    f"TRTR ACC Δ = "
    f"{trtr_acc.mean() - 0.8613:+.4f}"
)


print(
    f"TRTR AUC Δ = "
    f"{trtr_auc.mean() - 0.8745:+.4f}"
)




# ============================================================
# Fold seeds
# Useful for exact debugging / reproducibility
# ============================================================


print("\nFold generator seeds:")
print(fold_seeds)
```

# w/ GO-BS and GO-BS-FC Smoke test

```python
import os


# ============================================================
# Suppress warnings
# ============================================================


os.environ["PYTHONWARNINGS"] = "ignore"


import warnings
warnings.simplefilter("ignore")




# ============================================================
# Imports
# ============================================================


import numpy as np
import torch


from sklearn.base import clone


from bstabdiff import BSTabDiff




# ============================================================
# Configuration
# ============================================================


SEED = 0


N = 20
M_FEATURES = 100


N_BLOCKS = 10
N_SYN = 5


DEVICE = "cuda" if torch.cuda.is_available() else "cpu"




# ============================================================
# Synthetic smoke-test dataset
# ============================================================


rng = np.random.default_rng(SEED)


X = rng.standard_normal(
    (N, M_FEATURES)
).astype(np.float32)


y = rng.integers(
    0,
    2,
    size=N,
)




print("=" * 72)
print("BSTabDiff GO-BS / GO-BS-FC smoke tests")
print("=" * 72)


print("Device :", DEVICE)
print("X shape:", X.shape)
print("y shape:", y.shape)




# ============================================================
# Helper
# ============================================================


def run_smoke_test(
    name,
    use_gobs,
    use_gobs_fc,
):
    print("\n" + "=" * 72)
    print(name)
    print("=" * 72)


    model = BSTabDiff(
        M=N_BLOCKS,


        prior_type="diffusion",


        device=DEVICE,


        random_state=SEED,


        # ----------------------------------------------------
        # Tiny training configuration for smoke testing only
        # ----------------------------------------------------
        prior_epochs=5,
        prior_batch=16,
        prior_lr=1e-3,


        verbose_every=1,


        save_dir=None,
        save_best=True,


        use_ema=True,
        ema_decay=0.999,


        # ----------------------------------------------------
        # GO-BS settings
        # ----------------------------------------------------
        use_gobs=use_gobs,
        use_gobs_fc=use_gobs_fc,


        # Keep cluster count small because n=20.
        gobs_num_clusters=3,


        gobs_metric="kl_divergence",
        gobs_bins=32,


        gobs_top_k=None,


        gobs_refine_order=True,
        gobs_direction_select=True,
        gobs_refine_passes=1,


        gobs_boundary_refine_passes=5,
        gobs_boundary_window=8,


        gobs_lambda_cross=1.0,
        gobs_gamma_balance=1e-3,


        # False tests the GPU KMeans path when CUDA is available.
        gobs_use_cpu_kmeans=False,
    )




    # ========================================================
    # sklearn-style fit
    # ========================================================


    returned = model.fit(
        X,
        y,
    )




    # ========================================================
    # sklearn API checks
    # ========================================================


    assert returned is model


    assert hasattr(
        model,
        "generator_",
    )


    assert hasattr(
        model,
        "train_info_",
    )


    assert model.n_features_in_ == M_FEATURES




    # ========================================================
    # Generate synthetic samples
    # ========================================================


    X_syn, R_syn, y_syn = model.sample(
        n_samples=N_SYN,
        return_mask=True,
    )




    # ========================================================
    # Output checks
    # ========================================================


    assert X_syn.shape == (
        N_SYN,
        M_FEATURES,
    )


    assert R_syn.shape == (
        N_SYN,
        M_FEATURES,
    )


    assert y_syn is not None


    assert y_syn.shape == (
        N_SYN,
    )


    assert np.isfinite(
        X_syn[
            R_syn.astype(bool)
        ]
    ).all()




    # ========================================================
    # Check sklearn get_params()
    # ========================================================


    params = model.get_params()


    assert params["M"] == N_BLOCKS


    assert params["use_gobs"] == use_gobs


    assert params["use_gobs_fc"] == use_gobs_fc




    # ========================================================
    # Check sklearn clone()
    #
    # This clones configuration only, not fitted weights.
    # ========================================================


    cloned = clone(model)


    assert cloned.M == model.M


    assert (
        cloned.use_gobs
        == model.use_gobs
    )


    assert (
        cloned.use_gobs_fc
        == model.use_gobs_fc
    )




    # ========================================================
    # Report
    # ========================================================


    print()
    print("Smoke test OK")
    print("Synthetic X :", X_syn.shape)
    print("Synthetic R :", R_syn.shape)
    print("Synthetic y :", y_syn.shape)


    print(
        "Observed frac:",
        f"{R_syn.mean():.6f}",
    )


    print(
        "use_gobs     :",
        model.use_gobs,
    )


    print(
        "use_gobs_fc  :",
        model.use_gobs_fc,
    )


    print(
        "Device       :",
        model.device_,
    )


    return model




# ============================================================
# 1. GO-BS
# ============================================================


model_gobs = run_smoke_test(
    name="BSTabDiff + GO-BS",
    use_gobs=True,
    use_gobs_fc=False,
)




# ============================================================
# Cleanup between GPU tests
# ============================================================


if torch.cuda.is_available():
    torch.cuda.empty_cache()




# ============================================================
# 2. GO-BS-FC
# ============================================================


model_gobs_fc = run_smoke_test(
    name="BSTabDiff + GO-BS-FC",
    use_gobs=True,
    use_gobs_fc=True,
)




# ============================================================
# Final result
# ============================================================


print("\n" + "=" * 72)
print("ALL BSTabDiff ordering smoke tests PASSED")
print("=" * 72)


print("✓ GO-BS")
print("✓ GO-BS-FC")
print("✓ sklearn fit()")
print("✓ sample()")
print("✓ return_mask=True")
print("✓ get_params()")
print("✓ clone()")
```

# w/ GO-BS and GO-BS-FC on Colon

```python
import os


# ============================================================
# Suppress warnings
# ============================================================


os.environ["PYTHONWARNINGS"] = "ignore"


import warnings
warnings.simplefilter("ignore")




# ============================================================
# Imports
# ============================================================


import random
import numpy as np
import pandas as pd
import torch


from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state


from bstabdiff import BSTabDiff




# ============================================================
# Global experiment configuration
# ============================================================


GLOBAL_SEED = 42


# Exact strict 5x5 evaluation seed
CV_SEED = 0


CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"


N_SPLITS = 5
N_REPEATS = 5


DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"




# ============================================================
# Tuned Colon BSTabDiff configuration
#
# Paper:
# M       = 8
# prior   = flow
# epochs  = 300
# batch   = 64
# lr      = 3.2931e-3
# EMA     = 0.9926
# n_syn   = 64
# ============================================================


M = 8


PRIOR_TYPE = "flow"


PRIOR_EPOCHS = 300
PRIOR_BATCH = 64
PRIOR_LR = 3.2931e-3


EMA_DECAY = 0.9926


N_SYN = 64




# ============================================================
# Tuned Colon GO-BS configuration
#
# Paper GO-BS Optuna row:
#
# clusters          = 3
# metric            = correlation
# bins              = 24
# top-k             = 16
# refine order      = False
# direction select  = False
# refine passes     = 3
# boundary passes   = 5
# boundary window   = 16
# lambda_cross      = 2.3989
# gamma_balance     = 5.19e-6
# ============================================================


GOBS_NUM_CLUSTERS = 3
GOBS_METRIC = "correlation"
GOBS_BINS = 24
GOBS_TOP_K = 16


GOBS_REFINE_ORDER = False
GOBS_DIRECTION_SELECT = False
GOBS_REFINE_PASSES = 3


GOBS_BOUNDARY_REFINE_PASSES = 5
GOBS_BOUNDARY_WINDOW = 16


GOBS_LAMBDA_CROSS = 2.3989
GOBS_GAMMA_BALANCE = 5.19e-6




# ============================================================
# Paper reference values
# Strict 5x5 Logistic Regression
# ============================================================


PAPER_RESULTS = {
    "GO-BS": {
        "tstr_acc": 0.8287,
        "tstr_auc": 0.8878,
        "tstr_acc_std": 0.0994,
        "tstr_auc_std": 0.0961,
    },


    "GO-BS-FC": {
        "tstr_acc": 0.8321,
        "tstr_auc": 0.8935,
        "tstr_acc_std": 0.1009,
        "tstr_auc_std": 0.0917,
    },


    "TRTR": {
        "acc": 0.8613,
        "auc": 0.8745,
        "acc_std": 0.0856,
        "auc_std": 0.1056,
    },
}




# ============================================================
# Reproducibility
# ============================================================


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)


    torch.manual_seed(seed)


    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)




set_seed(GLOBAL_SEED)




# ============================================================
# Load Colon
# Same loading logic as previous notebook experiment
# ============================================================


def load_colon_dataset(
    csv_path=CSV_PATH,
    label_col=LABEL_COL,
):
    df = pd.read_csv(csv_path)


    if label_col not in df.columns:
        raise ValueError(
            f"Label column '{label_col}' not found. "
            f"Columns: {list(df.columns)[:10]} ..."
        )


    y_raw = df[label_col].values


    X_df = df.drop(
        columns=[label_col]
    )


    # Colon is numerical
    X_df = X_df.select_dtypes(
        include=[np.number]
    )


    X = X_df.to_numpy(
        dtype=np.float32
    )


    X = np.nan_to_num(
        X,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )


    # Encode labels as 0,...,C-1
    if y_raw.dtype.kind in {
        "U",
        "S",
        "O",
    }:
        y, _ = pd.factorize(
            y_raw
        )


    else:
        unique = np.unique(
            y_raw
        )


        if not np.array_equal(
            unique,
            np.arange(
                unique.size
            ),
        ):
            y, _ = pd.factorize(
                y_raw
            )


        else:
            y = y_raw.astype(
                int,
                copy=False,
            )


    return (
        X,
        np.asarray(
            y,
            dtype=int,
        ),
        list(
            X_df.columns
        ),
    )




# ============================================================
# Load dataset
# ============================================================


X, y, feature_names = load_colon_dataset()


print("=" * 76)


print(
    f"Loaded Colon: "
    f"X.shape={X.shape}, "
    f"y.shape={y.shape}, "
    f"classes={np.unique(y)}"
)


print(
    f"Device: {DEVICE}"
)


print("=" * 76)




assert X.shape == (62, 2000), (
    f"Expected Colon shape "
    f"(62, 2000), "
    f"got {X.shape}"
)




# ============================================================
# Logistic Regression
# Exact downstream configuration
# ============================================================


def make_lr():
    return Pipeline(
        [
            (
                "imp",
                SimpleImputer(
                    strategy="median",
                ),
            ),


            (
                "sc",
                StandardScaler(
                    with_mean=True,
                    with_std=True,
                ),
            ),


            (
                "clf",
                LogisticRegression(
                    max_iter=10000,
                    class_weight="balanced",
                    solver="lbfgs",
                    n_jobs=None,
                ),
            ),
        ]
    )




# ============================================================
# Evaluate one ordering method
# ============================================================


def evaluate_ordering(
    method_name,
    use_gobs_fc,
):
    """
    Strict 5x5 repeated stratified CV.


    GO-BS:
        use_gobs=True
        use_gobs_fc=False


    GO-BS-FC:
        use_gobs=True
        use_gobs_fc=True


    Both methods use identical:
        - folds
        - fold seeds
        - BSTabDiff hyperparameters
        - GO-BS hyperparameters
        - Logistic Regression evaluator
    """


    # --------------------------------------------------------
    # Reset global seed for clean method-level reproducibility
    # --------------------------------------------------------


    set_seed(
        GLOBAL_SEED
    )




    # --------------------------------------------------------
    # IMPORTANT:
    # reset RNG independently for every method so GO-BS and
    # GO-BS-FC use exactly the same fold seeds.
    # --------------------------------------------------------


    rng = check_random_state(
        CV_SEED
    )


    rkf = RepeatedStratifiedKFold(
        n_splits=N_SPLITS,
        n_repeats=N_REPEATS,
        random_state=CV_SEED,
    )




    tstr_acc = []
    tstr_auc = []


    trtr_acc = []
    trtr_auc = []


    fold_seeds = []




    print("\n\n" + "#" * 76)
    print(f"# {method_name}")
    print("#" * 76)


    print(
        f"M={M} | "
        f"prior={PRIOR_TYPE} | "
        f"epochs={PRIOR_EPOCHS} | "
        f"batch={PRIOR_BATCH} | "
        f"lr={PRIOR_LR} | "
        f"EMA={EMA_DECAY} | "
        f"n_syn={N_SYN}"
    )


    print(
        f"GO-BS clusters={GOBS_NUM_CLUSTERS} | "
        f"metric={GOBS_METRIC} | "
        f"bins={GOBS_BINS} | "
        f"top_k={GOBS_TOP_K}"
    )


    print(
        f"use_gobs=True | "
        f"use_gobs_fc={use_gobs_fc}"
    )


    print("#" * 76)




    # ========================================================
    # 25 folds
    # ========================================================


    for fold_id, (tr, te) in enumerate(
        rkf.split(X, y),
        start=1,
    ):


        # ----------------------------------------------------
        # Real train/test data
        # ----------------------------------------------------


        Xtr = X[tr]
        ytr = y[tr]


        Xte = X[te]
        yte = y[te]




        # ----------------------------------------------------
        # Exact fold seed generation
        # ----------------------------------------------------


        fold_seed = int(
            rng.randint(
                0,
                2**31 - 1,
            )
        )


        fold_seeds.append(
            fold_seed
        )




        print(
            "\n"
            + "=" * 76
        )


        print(
            f"{method_name} | "
            f"Fold {fold_id:02d}/25 | "
            f"train={len(tr)} | "
            f"test={len(te)} | "
            f"seed={fold_seed}"
        )


        print(
            "=" * 76
        )




        # ====================================================
        # BSTabDiff
        # ====================================================


        model = BSTabDiff(


            # ------------------------------------------------
            # Feature configuration
            # Colon contains numerical features only.
            # feature_specs=None lets wrapper create
            # continuous FeatureSpec entries automatically.
            # ------------------------------------------------


            feature_specs=None,


            M=M,


            blocks=None,


            # Learned GO-BS ordering handles permutation.
            # Do not add random feature permutation.
            permute_features=False,




            # ------------------------------------------------
            # Tuned Colon generator configuration
            # ------------------------------------------------


            prior_type=PRIOR_TYPE,


            device=DEVICE,


            random_state=fold_seed,


            prior_epochs=PRIOR_EPOCHS,


            prior_batch=PRIOR_BATCH,


            prior_lr=PRIOR_LR,


            verbose_every=0,


            save_dir=None,


            save_name=(
                "bstabdiff_colon_gobs_fc"
                if use_gobs_fc
                else
                "bstabdiff_colon_gobs"
            ),


            save_best=True,


            use_ema=True,


            ema_decay=EMA_DECAY,




            # ------------------------------------------------
            # Enable GO-BS
            # ------------------------------------------------


            use_gobs=True,


            # False = sample-clustered GO-BS
            # True  = feature-clustered GO-BS-FC
            use_gobs_fc=use_gobs_fc,




            # ------------------------------------------------
            # Tuned Colon GO-BS configuration
            # ------------------------------------------------


            gobs_num_clusters=GOBS_NUM_CLUSTERS,


            gobs_metric=GOBS_METRIC,


            gobs_bins=GOBS_BINS,


            gobs_top_k=GOBS_TOP_K,


            gobs_refine_order=GOBS_REFINE_ORDER,


            gobs_direction_select=GOBS_DIRECTION_SELECT,


            gobs_refine_passes=GOBS_REFINE_PASSES,


            gobs_boundary_refine_passes=(
                GOBS_BOUNDARY_REFINE_PASSES
            ),


            gobs_boundary_window=(
                GOBS_BOUNDARY_WINDOW
            ),


            gobs_lambda_cross=(
                GOBS_LAMBDA_CROSS
            ),


            gobs_gamma_balance=(
                GOBS_GAMMA_BALANCE
            ),


            # Implementation default:
            # use CUDA KMeans where supported.
            gobs_use_cpu_kmeans=False,
        )




        # ====================================================
        # Fit generator on TRAINING FOLD ONLY
        # ====================================================


        model.fit(
            Xtr,
            ytr,
        )




        # ====================================================
        # Generate tuned n_syn = 64
        # ====================================================


        X_syn, R_syn, y_syn = model.sample(
            n_samples=N_SYN,
            return_mask=True,
        )




        # ----------------------------------------------------
        # Sanity checks
        # ----------------------------------------------------


        assert X_syn.shape == (
            N_SYN,
            X.shape[1],
        )


        assert R_syn.shape == (
            N_SYN,
            X.shape[1],
        )


        assert y_syn is not None


        assert y_syn.shape == (
            N_SYN,
        )




        # ====================================================
        # TSTR
        #
        # Train on synthetic
        # Test on real held-out fold
        # ====================================================


        clf_tstr = make_lr()


        clf_tstr.fit(
            X_syn,
            y_syn,
        )


        proba_tstr = (
            clf_tstr
            .predict_proba(
                Xte
            )[:, 1]
        )


        pred_tstr = (
            proba_tstr >= 0.5
        ).astype(int)




        fold_tstr_acc = (
            accuracy_score(
                yte,
                pred_tstr,
            )
        )


        fold_tstr_auc = (
            roc_auc_score(
                yte,
                proba_tstr,
            )
        )




        tstr_acc.append(
            fold_tstr_acc
        )


        tstr_auc.append(
            fold_tstr_auc
        )




        # ====================================================
        # TRTR
        #
        # Train on real
        # Test on real
        # ====================================================


        clf_trtr = make_lr()


        clf_trtr.fit(
            Xtr,
            ytr,
        )


        proba_trtr = (
            clf_trtr
            .predict_proba(
                Xte
            )[:, 1]
        )


        pred_trtr = (
            proba_trtr >= 0.5
        ).astype(int)




        fold_trtr_acc = (
            accuracy_score(
                yte,
                pred_trtr,
            )
        )


        fold_trtr_auc = (
            roc_auc_score(
                yte,
                proba_trtr,
            )
        )




        trtr_acc.append(
            fold_trtr_acc
        )


        trtr_auc.append(
            fold_trtr_auc
        )




        # ====================================================
        # Fold report
        # ====================================================


        print(
            f"TSTR | "
            f"ACC={fold_tstr_acc:.4f} | "
            f"AUC={fold_tstr_auc:.4f}"
        )


        print(
            f"TRTR | "
            f"ACC={fold_trtr_acc:.4f} | "
            f"AUC={fold_trtr_auc:.4f}"
        )




        # ====================================================
        # Cleanup
        # ====================================================


        del model
        del clf_tstr
        del clf_trtr


        if torch.cuda.is_available():
            torch.cuda.empty_cache()




    # ========================================================
    # Convert to NumPy
    # ========================================================


    tstr_acc = np.asarray(
        tstr_acc,
        dtype=float,
    )


    tstr_auc = np.asarray(
        tstr_auc,
        dtype=float,
    )


    trtr_acc = np.asarray(
        trtr_acc,
        dtype=float,
    )


    trtr_auc = np.asarray(
        trtr_auc,
        dtype=float,
    )




    # ========================================================
    # Results
    # ========================================================


    result = {
        "method": method_name,


        "tstr_acc_mean": tstr_acc.mean(),
        "tstr_acc_std": tstr_acc.std(),


        "tstr_auc_mean": tstr_auc.mean(),
        "tstr_auc_std": tstr_auc.std(),


        "trtr_acc_mean": trtr_acc.mean(),
        "trtr_acc_std": trtr_acc.std(),


        "trtr_auc_mean": trtr_auc.mean(),
        "trtr_auc_std": trtr_auc.std(),


        "fold_seeds": fold_seeds,
    }




    # ========================================================
    # Summary
    # ========================================================


    print(
        "\n\n"
        + "=" * 76
    )


    print(
        f"COLON — {method_name}"
    )


    print(
        "STRICT 5 x 5 repeated stratified CV"
    )


    print(
        f"GLOBAL_SEED = {GLOBAL_SEED}"
    )


    print(
        f"CV_SEED     = {CV_SEED}"
    )


    print(
        f"Device      = {DEVICE}"
    )


    print(
        "=" * 76
    )




    print("\nTSTR:")


    print(
        f"ACC = "
        f"{result['tstr_acc_mean']:.4f} "
        f"± "
        f"{result['tstr_acc_std']:.4f}"
    )


    print(
        f"AUC = "
        f"{result['tstr_auc_mean']:.4f} "
        f"± "
        f"{result['tstr_auc_std']:.4f}"
    )




    print("\nTRTR:")


    print(
        f"ACC = "
        f"{result['trtr_acc_mean']:.4f} "
        f"± "
        f"{result['trtr_acc_std']:.4f}"
    )


    print(
        f"AUC = "
        f"{result['trtr_auc_mean']:.4f} "
        f"± "
        f"{result['trtr_auc_std']:.4f}"
    )




    # ========================================================
    # Paper reference
    # ========================================================


    ref = PAPER_RESULTS[
        method_name
    ]


    print(
        "\n"
        + "-" * 76
    )


    print(
        "Paper reference"
    )


    print(
        "-" * 76
    )


    print(
        f"TSTR ACC = "
        f"{ref['tstr_acc']:.4f} "
        f"± "
        f"{ref['tstr_acc_std']:.4f}"
    )


    print(
        f"TSTR AUC = "
        f"{ref['tstr_auc']:.4f} "
        f"± "
        f"{ref['tstr_auc_std']:.4f}"
    )


    print(
        f"TRTR ACC = "
        f"{PAPER_RESULTS['TRTR']['acc']:.4f} "
        f"± "
        f"{PAPER_RESULTS['TRTR']['acc_std']:.4f}"
    )


    print(
        f"TRTR AUC = "
        f"{PAPER_RESULTS['TRTR']['auc']:.4f} "
        f"± "
        f"{PAPER_RESULTS['TRTR']['auc_std']:.4f}"
    )




    # ========================================================
    # Difference from paper
    # ========================================================


    print(
        "\n"
        + "-" * 76
    )


    print(
        "Difference: current run - paper"
    )


    print(
        "-" * 76
    )


    print(
        f"TSTR ACC Δ = "
        f"{result['tstr_acc_mean'] - ref['tstr_acc']:+.4f}"
    )


    print(
        f"TSTR AUC Δ = "
        f"{result['tstr_auc_mean'] - ref['tstr_auc']:+.4f}"
    )


    print(
        f"TRTR ACC Δ = "
        f"{result['trtr_acc_mean'] - PAPER_RESULTS['TRTR']['acc']:+.4f}"
    )


    print(
        f"TRTR AUC Δ = "
        f"{result['trtr_auc_mean'] - PAPER_RESULTS['TRTR']['auc']:+.4f}"
    )




    print(
        "\nFold generator seeds:"
    )


    print(
        fold_seeds
    )


    return result




# ============================================================
# Run 1: GO-BS
# ============================================================


result_gobs = evaluate_ordering(
    method_name="GO-BS",
    use_gobs_fc=False,
)




# ============================================================
# Clean GPU before second full experiment
# ============================================================


if torch.cuda.is_available():
    torch.cuda.empty_cache()




# ============================================================
# Run 2: GO-BS-FC
# ============================================================


result_gobs_fc = evaluate_ordering(
    method_name="GO-BS-FC",
    use_gobs_fc=True,
)




# ============================================================
# Combined comparison
# ============================================================


summary = pd.DataFrame(
    [
        {
            "Method": "GO-BS",


            "TSTR ACC":
                result_gobs[
                    "tstr_acc_mean"
                ],


            "TSTR ACC std":
                result_gobs[
                    "tstr_acc_std"
                ],


            "TSTR AUC":
                result_gobs[
                    "tstr_auc_mean"
                ],


            "TSTR AUC std":
                result_gobs[
                    "tstr_auc_std"
                ],


            "TRTR ACC":
                result_gobs[
                    "trtr_acc_mean"
                ],


            "TRTR AUC":
                result_gobs[
                    "trtr_auc_mean"
                ],
        },


        {
            "Method": "GO-BS-FC",


            "TSTR ACC":
                result_gobs_fc[
                    "tstr_acc_mean"
                ],


            "TSTR ACC std":
                result_gobs_fc[
                    "tstr_acc_std"
                ],


            "TSTR AUC":
                result_gobs_fc[
                    "tstr_auc_mean"
                ],


            "TSTR AUC std":
                result_gobs_fc[
                    "tstr_auc_std"
                ],


            "TRTR ACC":
                result_gobs_fc[
                    "trtr_acc_mean"
                ],


            "TRTR AUC":
                result_gobs_fc[
                    "trtr_auc_mean"
                ],
        },
    ]
)




print(
    "\n\n"
    + "=" * 90
)


print(
    "FINAL COLON GO-BS / GO-BS-FC COMPARISON"
)


print(
    "=" * 90
)


print(
    summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}",
    )
)




# ============================================================
# Verify both methods used exactly the same fold seeds
# ============================================================


assert (
    result_gobs["fold_seeds"]
    ==
    result_gobs_fc["fold_seeds"]
), (
    "GO-BS and GO-BS-FC did not use "
    "identical generator seeds."
)




print(
    "\n✓ GO-BS and GO-BS-FC used identical "
    "CV splits and fold generator seeds."
)




# ============================================================
# Optional: save summary
# ============================================================


summary.to_csv(
    "colon_gobs_gobsfc_reproduction.csv",
    index=False,
)


print(
    "✓ Saved: "
    "colon_gobs_gobsfc_reproduction.csv"
)
```

```python
import os


# ============================================================
# Suppress warnings
# ============================================================


os.environ["PYTHONWARNINGS"] = "ignore"


import warnings
warnings.simplefilter("ignore")




# ============================================================
# Imports
# ============================================================


import random
import numpy as np
import pandas as pd
import torch


from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import check_random_state


from bstabdiff import BSTabDiff




# ============================================================
# General experiment configuration
# ============================================================


GLOBAL_SEED = 42


# Same strict 5x5 evaluation seed used for Colon
CV_SEED = 0


CSV_PATH = "coloncancer_encoded.csv"
LABEL_COL = "label"


N_SPLITS = 5
N_REPEATS = 5


DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"




# ============================================================
# OPTUNA-TUNED COLON CONFIGURATION
#
# No GO-BS
# No feature permutation
#
# M          = 8
# prior      = flow
# epochs     = 300
# batch      = 64
# lr         = 3.2931e-3
# EMA        = 0.9926
# n_syn      = 64
# ============================================================


M = 8


PRIOR_TYPE = "flow"


PRIOR_EPOCHS = 300
PRIOR_BATCH = 64
PRIOR_LR = 3.2931e-3


EMA_DECAY = 0.9926


N_SYN = 64




# ============================================================
# Paper reference
#
# Main LR table:
# BSTabDiff tuned, no permutation:
# COL TSTR ACC = 0.8321 ± 0.0981
#
# Real TRTR:
# ACC = 0.8613 ± 0.0856
#
# We will also calculate AUC here.
# ============================================================


PAPER_TSTR_ACC = 0.8321
PAPER_TSTR_ACC_STD = 0.0981


PAPER_TRTR_ACC = 0.8613
PAPER_TRTR_ACC_STD = 0.0856


# From the same strict Colon LR folds already reproduced
PAPER_TRTR_AUC = 0.8745
PAPER_TRTR_AUC_STD = 0.1056




# ============================================================
# Reproducibility
# ============================================================


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)


    torch.manual_seed(seed)


    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)




set_seed(GLOBAL_SEED)




# ============================================================
# Load Colon dataset
# ============================================================


def load_colon_dataset(
    csv_path=CSV_PATH,
    label_col=LABEL_COL,
):
    df = pd.read_csv(csv_path)


    if label_col not in df.columns:
        raise ValueError(
            f"Label column '{label_col}' not found. "
            f"Columns: {list(df.columns)[:10]} ..."
        )


    y_raw = df[label_col].values


    X_df = df.drop(
        columns=[label_col]
    )


    # Colon contains numerical features
    X_df = X_df.select_dtypes(
        include=[np.number]
    )


    X = X_df.to_numpy(
        dtype=np.float32
    )


    # Same preprocessing as previous Colon experiments
    X = np.nan_to_num(
        X,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )


    # Encode labels to 0,...,C-1
    if y_raw.dtype.kind in {"U", "S", "O"}:


        y, _ = pd.factorize(
            y_raw
        )


    else:


        unique = np.unique(
            y_raw
        )


        if not np.array_equal(
            unique,
            np.arange(unique.size),
        ):


            y, _ = pd.factorize(
                y_raw
            )


        else:


            y = y_raw.astype(
                int,
                copy=False,
            )


    return (
        X,
        np.asarray(
            y,
            dtype=int,
        ),
        list(X_df.columns),
    )




# ============================================================
# Load data
# ============================================================


X, y, feature_names = load_colon_dataset()


print("=" * 76)


print(
    f"Loaded Colon dataset: "
    f"X.shape={X.shape}, "
    f"y.shape={y.shape}, "
    f"classes={np.unique(y)}"
)


print(
    f"Device: {DEVICE}"
)


print("=" * 76)




assert X.shape == (62, 2000), (
    f"Expected Colon shape (62, 2000), "
    f"got {X.shape}"
)




# ============================================================
# Logistic Regression
# Same evaluator as paper/notebook
# ============================================================


def make_lr():


    return Pipeline(
        [
            (
                "imp",
                SimpleImputer(
                    strategy="median",
                ),
            ),


            (
                "sc",
                StandardScaler(
                    with_mean=True,
                    with_std=True,
                ),
            ),


            (
                "clf",
                LogisticRegression(
                    max_iter=10000,
                    class_weight="balanced",
                    solver="lbfgs",
                    n_jobs=None,
                ),
            ),
        ]
    )




# ============================================================
# Strict 5x5 repeated stratified CV
# ============================================================


rng = check_random_state(
    CV_SEED
)


rkf = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=CV_SEED,
)




# ============================================================
# Result storage
# ============================================================


tstr_acc = []
tstr_auc = []


trtr_acc = []
trtr_auc = []


fold_seeds = []




# ============================================================
# Main evaluation
# ============================================================


for fold_id, (tr, te) in enumerate(
    rkf.split(X, y),
    start=1,
):


    # --------------------------------------------------------
    # Train/test fold
    # --------------------------------------------------------


    Xtr = X[tr]
    ytr = y[tr]


    Xte = X[te]
    yte = y[te]




    # --------------------------------------------------------
    # Same deterministic fold-seed generation
    # --------------------------------------------------------


    fold_seed = int(
        rng.randint(
            0,
            2**31 - 1,
        )
    )


    fold_seeds.append(
        fold_seed
    )




    print(
        "\n"
        + "=" * 76
    )


    print(
        f"Fold {fold_id:02d}/25 | "
        f"train={len(tr)} | "
        f"test={len(te)} | "
        f"seed={fold_seed}"
    )


    print(
        "=" * 76
    )




    # ========================================================
    # 1. Tuned BSTabDiff
    #
    # IMPORTANT:
    #
    # use_gobs=False
    # use_gobs_fc=False
    # permute_features=False
    #
    # This is the tuned NO-PERMUTATION configuration.
    # ========================================================


    model = BSTabDiff(


        # Colon features are continuous
        feature_specs=None,


        # Tuned block count
        M=M,


        # No predefined custom blocks
        blocks=None,


        # ----------------------------------------------------
        # NO permutation
        # ----------------------------------------------------


        permute_features=False,


        # ----------------------------------------------------
        # Tuned generator
        # ----------------------------------------------------


        prior_type=PRIOR_TYPE,


        device=DEVICE,


        random_state=fold_seed,


        prior_epochs=PRIOR_EPOCHS,


        prior_batch=PRIOR_BATCH,


        prior_lr=PRIOR_LR,


        verbose_every=0,


        save_dir=None,


        save_name="bstabdiff_colon_tuned_noperm",


        save_best=True,


        use_ema=True,


        ema_decay=EMA_DECAY,


        # ----------------------------------------------------
        # Explicitly disable feature ordering
        # ----------------------------------------------------


        use_gobs=False,


        use_gobs_fc=False,
    )




    # ========================================================
    # Fit generator ONLY on current training fold
    # ========================================================


    model.fit(
        Xtr,
        ytr,
    )




    # ========================================================
    # Generate exactly 64 synthetic samples
    # ========================================================


    X_syn, R_syn, y_syn = model.sample(
        n_samples=N_SYN,
        return_mask=True,
    )




    # ========================================================
    # Sanity checks
    # ========================================================


    assert X_syn.shape == (
        N_SYN,
        X.shape[1],
    )


    assert R_syn.shape == (
        N_SYN,
        X.shape[1],
    )


    assert y_syn is not None


    assert y_syn.shape == (
        N_SYN,
    )




    # ========================================================
    # 2. TSTR
    #
    # Train on synthetic
    # Test on held-out real
    # ========================================================


    clf_tstr = make_lr()


    clf_tstr.fit(
        X_syn,
        y_syn,
    )




    proba_tstr = (
        clf_tstr
        .predict_proba(
            Xte
        )[:, 1]
    )




    pred_tstr = (
        proba_tstr >= 0.5
    ).astype(int)




    fold_tstr_acc = accuracy_score(
        yte,
        pred_tstr,
    )


    fold_tstr_auc = roc_auc_score(
        yte,
        proba_tstr,
    )




    tstr_acc.append(
        fold_tstr_acc
    )


    tstr_auc.append(
        fold_tstr_auc
    )




    # ========================================================
    # 3. TRTR
    #
    # Train on real
    # Test on real
    # ========================================================


    clf_trtr = make_lr()


    clf_trtr.fit(
        Xtr,
        ytr,
    )




    proba_trtr = (
        clf_trtr
        .predict_proba(
            Xte
        )[:, 1]
    )




    pred_trtr = (
        proba_trtr >= 0.5
    ).astype(int)




    fold_trtr_acc = accuracy_score(
        yte,
        pred_trtr,
    )


    fold_trtr_auc = roc_auc_score(
        yte,
        proba_trtr,
    )




    trtr_acc.append(
        fold_trtr_acc
    )


    trtr_auc.append(
        fold_trtr_auc
    )




    # ========================================================
    # Fold output
    # ========================================================


    print(
        f"TSTR | "
        f"ACC={fold_tstr_acc:.4f} | "
        f"AUC={fold_tstr_auc:.4f}"
    )


    print(
        f"TRTR | "
        f"ACC={fold_trtr_acc:.4f} | "
        f"AUC={fold_trtr_auc:.4f}"
    )




    # ========================================================
    # Cleanup
    # ========================================================


    del model
    del clf_tstr
    del clf_trtr


    if torch.cuda.is_available():
        torch.cuda.empty_cache()




# ============================================================
# Convert to NumPy arrays
# ============================================================


tstr_acc = np.asarray(
    tstr_acc,
    dtype=float,
)


tstr_auc = np.asarray(
    tstr_auc,
    dtype=float,
)


trtr_acc = np.asarray(
    trtr_acc,
    dtype=float,
)


trtr_auc = np.asarray(
    trtr_auc,
    dtype=float,
)




# ============================================================
# Final summary
# ============================================================


print(
    "\n\n"
    + "=" * 76
)


print(
    "COLON — TUNED BSTabDiff, NO PERMUTATION"
)


print(
    "STRICT 5 x 5 repeated stratified CV"
)


print(
    f"GLOBAL_SEED = {GLOBAL_SEED}"
)


print(
    f"CV_SEED     = {CV_SEED}"
)


print(
    f"Device      = {DEVICE}"
)


print(
    f"M           = {M}"
)


print(
    f"Prior       = {PRIOR_TYPE}"
)


print(
    f"Epochs      = {PRIOR_EPOCHS}"
)


print(
    f"Batch       = {PRIOR_BATCH}"
)


print(
    f"LR          = {PRIOR_LR}"
)


print(
    f"EMA         = {EMA_DECAY}"
)


print(
    f"n_syn       = {N_SYN}"
)


print(
    "GO-BS       = False"
)


print(
    "Permutation = False"
)


print("=" * 76)




# ============================================================
# TSTR
# ============================================================


print("\nTSTR:")


print(
    f"ACC = "
    f"{tstr_acc.mean():.4f} "
    f"± {tstr_acc.std():.4f}"
)


print(
    f"AUC = "
    f"{tstr_auc.mean():.4f} "
    f"± {tstr_auc.std():.4f}"
)




# ============================================================
# TRTR
# ============================================================


print("\nTRTR:")


print(
    f"ACC = "
    f"{trtr_acc.mean():.4f} "
    f"± {trtr_acc.std():.4f}"
)


print(
    f"AUC = "
    f"{trtr_auc.mean():.4f} "
    f"± {trtr_auc.std():.4f}"
)




# ============================================================
# Paper reference
# ============================================================


print(
    "\n"
    + "-" * 76
)


print(
    "Paper reference — tuned BSTabDiff, no permutation"
)


print(
    "-" * 76
)


print(
    f"TSTR ACC = "
    f"{PAPER_TSTR_ACC:.4f} "
    f"± {PAPER_TSTR_ACC_STD:.4f}"
)


print(
    f"TRTR ACC = "
    f"{PAPER_TRTR_ACC:.4f} "
    f"± {PAPER_TRTR_ACC_STD:.4f}"
)


print(
    f"TRTR AUC = "
    f"{PAPER_TRTR_AUC:.4f} "
    f"± {PAPER_TRTR_AUC_STD:.4f}"
)




# ============================================================
# Difference from paper/reference
# ============================================================


print(
    "\n"
    + "-" * 76
)


print(
    "Difference: current run - reference"
)


print(
    "-" * 76
)


print(
    f"TSTR ACC Δ = "
    f"{tstr_acc.mean() - PAPER_TSTR_ACC:+.4f}"
)


print(
    f"TRTR ACC Δ = "
    f"{trtr_acc.mean() - PAPER_TRTR_ACC:+.4f}"
)


print(
    f"TRTR AUC Δ = "
    f"{trtr_auc.mean() - PAPER_TRTR_AUC:+.4f}"
)




# ============================================================
# Fold seeds
# ============================================================


print(
    "\nFold generator seeds:"
)


print(
    fold_seeds
)




# ============================================================
# Save result
# ============================================================


summary = pd.DataFrame(
    [
        {
            "Configuration":
                "BSTabDiff tuned no permutation",


            "M":
                M,


            "Prior":
                PRIOR_TYPE,


            "Epochs":
                PRIOR_EPOCHS,


            "Batch":
                PRIOR_BATCH,


            "LR":
                PRIOR_LR,


            "EMA":
                EMA_DECAY,


            "n_syn":
                N_SYN,


            "TSTR ACC":
                tstr_acc.mean(),


            "TSTR ACC std":
                tstr_acc.std(),


            "TSTR AUC":
                tstr_auc.mean(),


            "TSTR AUC std":
                tstr_auc.std(),


            "TRTR ACC":
                trtr_acc.mean(),


            "TRTR ACC std":
                trtr_acc.std(),


            "TRTR AUC":
                trtr_auc.mean(),


            "TRTR AUC std":
                trtr_auc.std(),
        }
    ]
)




summary.to_csv(
    "colon_bstabdiff_tuned_noperm_reproduction.csv",
    index=False,
)


print(
    "\n✓ Saved: "
    "colon_bstabdiff_tuned_noperm_reproduction.csv"
)
```
